## 6. Hierarchical Clustering

Compute and display a dendrogram using the linkage method `(10 pts)`
- Use `'ward'` linkage to minimize variance within clusters.

```python
linked = linkage(df_num_scaled, method='ward')
plt.figure(figsize=(18, 6))
dendrogram(linked)
plt.title('Dendrogram (Ward Linkage) for the clustering of the dataset')
plt.xlabel("Samples")
plt.ylabel("Distance")
plt.show()
```

Based on the dendrogram, choose a reasonable number of clusters `(5 pts)`
- Write your chosen number below and explain your reasoning.

```python
'''
3 clusters because there is a large vertical jump at around 20–27 when the clustering reduces from three groups to two.
This big height difference means that merging beyond three clusters would combine groups that are not similar.
Before that point the merges occur at much smaller distances. Therefore, 3 clusters is the most reasonable choice.

'''
```

Perform Agglomerative Clustering using your chosen cluster count `(10 pts)`
- Attach the cluster labels back to the original dataframe.

```python
clustering_features = df_num_scaled[['income', 'gdpp']]

cluster_H = AgglomerativeClustering(n_clusters=3, linkage='ward')
model_clt = cluster_H.fit(clustering_features)

df_num['clusters'] = model_clt.labels_

print(model_clt)
print('\n')
print('Clusters assigned to each datapoint, cluster = 3 :')
print(df_num['clusters'])
```

## 7. Visualizing Cluster Results

Create a scatter plot using any 2 meaningful features, colored by cluster `(10 pts)`

```python
plt.scatter(df_num_scaled['gdpp'],
            df_num_scaled['income'],
            c=df_num_scaled['clusters'],
            cmap='inferno',
            s=10,
            )

plt.xlabel('gdpp')
plt.ylabel('Income')

plt.title('Scatter plot of income vs gdpp by Cluster')
plt.colorbar(label='Magma')

plt.show()
```
